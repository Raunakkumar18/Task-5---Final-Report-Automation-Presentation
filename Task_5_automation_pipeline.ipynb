```python
"""
Task 5 - Simple Sales Analytics Automation Pipeline
Loads raw Sample Superstore CSV, cleans data, calculates KPIs,
exports processed Excel, and creates KPI/visual reports.

Run:
    python automation_pipeline.ipynb
"""
```

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# ==========================================
# 1. PATH SETUP
# ==========================================

BASE = Path.cwd()

INPUT = BASE / "SampleSuperstore(5).csv"

OUTPUT = BASE / "output"
CHARTS = OUTPUT / "charts"

OUTPUT.mkdir(exist_ok=True)
CHARTS.mkdir(exist_ok=True)

print("Base folder:", BASE)
print("Input file:", INPUT)


# ==========================================
# 2. LOAD DATA
# ==========================================

df = pd.read_csv(INPUT)

print("Data loaded successfully!")
print("Rows:", len(df))
print("Columns:", len(df.columns))

display(df.head())


# ==========================================
# 3. CLEAN DATA
# ==========================================

df.columns = [
    str(col).replace("\ufeff", "").strip()
    for col in df.columns
]

# Convert dates
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    errors="coerce"
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    errors="coerce"
)

# Numeric columns
numeric_columns = [
    "Sales",
    "Quantity",
    "Discount",
    "Profit",
    "Postal Code"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

# Remove duplicates
df = df.drop_duplicates()

# Remove invalid essential records
df = df.dropna(
    subset=[
        "Order Date",
        "Order ID",
        "Sales",
        "Profit"
    ]
)

# Derived columns
df["Order Year"] = df["Order Date"].dt.year

df["Order Month"] = (
    df["Order Date"]
    .dt.to_period("M")
    .astype(str)
)

print("Cleaning completed!")
print("Clean rows:", len(df))


# ==========================================
# 4. KPI CALCULATION
# ==========================================

total_sales = df["Sales"].sum()

total_profit = df["Profit"].sum()

total_orders = df["Order ID"].nunique()

total_customers = df["Customer ID"].nunique()

total_quantity = df["Quantity"].sum()

average_order_value = (
    df.groupby("Order ID")["Sales"]
    .sum()
    .mean()
)

profit_margin = (
    total_profit / total_sales
    if total_sales != 0
    else 0
)

print("\n========== KPI SUMMARY ==========")

print(f"Total Sales       : ${total_sales:,.2f}")
print(f"Total Profit      : ${total_profit:,.2f}")
print(f"Total Orders      : {total_orders:,}")
print(f"Total Customers   : {total_customers:,}")
print(f"Total Quantity    : {total_quantity:,}")
print(f"Average Order     : ${average_order_value:,.2f}")
print(f"Profit Margin     : {profit_margin:.2%}")


# ==========================================
# 5. KPI DATAFRAME
# ==========================================

kpi = pd.DataFrame({
    "KPI": [
        "Total Sales",
        "Total Profit",
        "Total Orders",
        "Total Customers",
        "Total Quantity",
        "Average Order Value",
        "Profit Margin"
    ],
    "Value": [
        total_sales,
        total_profit,
        total_orders,
        total_customers,
        total_quantity,
        average_order_value,
        profit_margin
    ]
})

display(kpi)


# ==========================================
# 6. CATEGORY ANALYSIS
# ==========================================

category = (
    df.groupby("Category")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum")
    )
    .reset_index()
    .sort_values(
        "Sales",
        ascending=False
    )
)

display(category)


# ==========================================
# 7. SEGMENT ANALYSIS
# ==========================================

segment = (
    df.groupby("Segment")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum")
    )
    .reset_index()
    .sort_values(
        "Sales",
        ascending=False
    )
)

display(segment)


# ==========================================
# 8. REGION ANALYSIS
# ==========================================

region = (
    df.groupby("Region")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum")
    )
    .reset_index()
    .sort_values(
        "Sales",
        ascending=False
    )
)

display(region)


# ==========================================
# 9. TOP 10 PRODUCTS
# ==========================================

top_products = (
    df.groupby("Product Name")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum")
    )
    .reset_index()
    .sort_values(
        "Sales",
        ascending=False
    )
    .head(10)
)

display(top_products)


# ==========================================
# 10. MONTHLY SALES
# ==========================================

monthly = (
    df.groupby("Order Month")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum")
    )
    .reset_index()
)

display(monthly)


# ==========================================
# 11. YEARLY SALES
# ==========================================

yearly = (
    df.groupby("Order Year")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum")
    )
    .reset_index()
)

display(yearly)


# ==========================================
# 12. CHART 1 - MONTHLY SALES
# ==========================================

plt.figure(figsize=(12, 5))

plt.plot(
    monthly["Order Month"],
    monthly["Sales"],
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.xticks(rotation=60)

plt.tight_layout()

plt.savefig(
    CHARTS / "monthly_sales.png",
    dpi=160
)

plt.show()


# ==========================================
# 13. CHART 2 - CATEGORY SALES
# ==========================================

plt.figure(figsize=(8, 5))

plt.bar(
    category["Category"],
    category["Sales"]
)

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")

plt.tight_layout()

plt.savefig(
    CHARTS / "sales_category.png",
    dpi=160
)

plt.show()


# ==========================================
# 14. CHART 3 - SEGMENT SALES
# ==========================================

plt.figure(figsize=(8, 5))

plt.bar(
    segment["Segment"],
    segment["Sales"]
)

plt.title("Sales by Segment")
plt.xlabel("Segment")
plt.ylabel("Sales")

plt.tight_layout()

plt.savefig(
    CHARTS / "sales_segment.png",
    dpi=160
)

plt.show()


# ==========================================
# 15. CHART 4 - TOP PRODUCTS
# ==========================================

top5 = top_products.head(5).sort_values(
    "Sales"
)

plt.figure(figsize=(9, 5))

plt.barh(
    top5["Product Name"],
    top5["Sales"]
)

plt.title("Top 5 Products by Sales")
plt.xlabel("Sales")

plt.tight_layout()

plt.savefig(
    CHARTS / "top_products.png",
    dpi=160
)

plt.show()


# ==========================================
# 16. EXPORT TO EXCEL
# ==========================================

excel_file = OUTPUT / "sales_analytics_output.xlsx"

with pd.ExcelWriter(
    excel_file,
    engine="openpyxl"
) as writer:

    df.to_excel(
        writer,
        sheet_name="Processed_Data",
        index=False
    )

    kpi.to_excel(
        writer,
        sheet_name="KPI_Summary",
        index=False
    )

    category.to_excel(
        writer,
        sheet_name="Category",
        index=False
    )

    segment.to_excel(
        writer,
        sheet_name="Segment",
        index=False
    )

    region.to_excel(
        writer,
        sheet_name="Region",
        index=False
    )

    top_products.to_excel(
        writer,
        sheet_name="Top_Products",
        index=False
    )

    monthly.to_excel(
        writer,
        sheet_name="Monthly",
        index=False
    )

    yearly.to_excel(
        writer,
        sheet_name="Yearly",
        index=False
    )

print("Excel file created:")
print(excel_file)


# ==========================================
# 17. KPI TEXT FILE
# ==========================================

summary = f"""
TASK 5 - SALES ANALYTICS SUMMARY

Total Sales       : ${total_sales:,.2f}
Total Profit      : ${total_profit:,.2f}
Total Orders      : {total_orders:,}
Total Customers   : {total_customers:,}
Total Quantity    : {total_quantity:,}
Average Order     : ${average_order_value:,.2f}
Profit Margin     : {profit_margin:.2%}

Pipeline Status: COMPLETED
"""

summary_file = OUTPUT / "kpi_summary.txt"

summary_file.write_text(
    summary,
    encoding="utf-8"
)

print(summary)

print("================================")
print("TASK 5 AUTOMATION COMPLETED")
print("================================")
print("Output folder:", OUTPUT)
```
